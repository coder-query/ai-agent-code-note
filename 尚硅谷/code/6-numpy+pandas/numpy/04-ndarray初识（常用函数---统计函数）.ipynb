```python

# np.sum()	所有元素的和
# np.max()	所有元素的最大值
# np.min()	所有元素的最小值
# np.std()	所有元素的标准差
# np.var()	所有元素的方差
# np.argmax()	最大值的下标索引值
# np.argmin()	最小值的下标索引值
# np.cumsum()	返回一个一维数组，每个元素都是之前所有元素的累加和
# np.cumprod()	返回一个一维数组，每个元素都是之前所有元素的累乘积

# 多维数组在计算时默认计算全部维度，可以使用axis参数指定按某一维度为轴心统计，axis=0按列统计、axis=1按行统计
import  numpy as np
arr1 = np.random.randint(1, 5, (2, 3))
print(arr1)
# np.mean()	所有元素的平均值
print(np.mean(arr1))
# np.sum()	所有元素的和
print(np.sum(arr1))
# np.max()	所有元素的最大值
print(np.max(arr1))
# np.min()	所有元素的最小值
print(np.min(arr1))
# np.std()	所有元素的标准差
print(np.std(arr1))
# np.var()	所有元素的方差
print(np.var(arr1))
# np.argmax()	最大值的下标索引值
print(np.argmax(arr1))
# np.argmin()	最小值的下标索引值
print(np.argmin(arr1))
# np.cumsum()	返回一个一维数组，每个元素都是之前所有元素的累加和
print(np.cumsum(arr1))
# np.cumprod()	返回一个一维数组，每个元素都是之前所有元素的累乘积
print(np.cumprod(arr1))
print(np.cumprod(arr1, axis=1))
```
